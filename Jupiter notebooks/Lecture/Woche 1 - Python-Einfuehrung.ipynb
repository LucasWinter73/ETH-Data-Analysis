### Warum Python?

* Mit Python kann man schnell und interaktiv arbeiten

````python
import matplotlib.pyplot as plt

x, y = 0, 0
vx, vy = 15, 20  
g, dt = 9.81, 0.05          
x_hist, y_hist = [0.0], [0.0]

while y >= 0:
    vy += -g * dt             
    x += vx * dt             
    y += vy * dt             
    x_hist.append(x); y_hist.append(y)

plt.plot(x_hist, y_hist, 'ro-') 
plt.show()
````

### Ein erstes Python Programm

```python 
def compute_mean(measurements):
    """Compute mean of a list of measurements."""
    
    if len(measurements) == 0:
        return 0 # avoid division by zero
    
    sum = 0
    for value in measurements:
        sum += value
    mean = sum / len(measurements) 
    return mean

measurements = [1, 0.1, 2.0]
compute_mean(measurements)
```

### Funktionen 

* Funktionen werden mit `def` deklariert
* Mehrere Rückgabewerte mit Tuple

````python
def min_max(values):
    min_value = min(values)
    max_value = max(values)
    return (min_value, max_value)

(minimum, maximum) = min_max([3, 5, 1, 9, 2])
print(minimum)
print(maximum)
````

### Plotten mit Matplotlib

* Matplotlib stellt uns Funktionen zum Erstellen von Diagrammen zur Verfügung

````python
import matplotlib.pyplot as plt

x = np.linspace(0, 10, 100)
y = np.sin(x)

plt.plot(x, y)
plt.title("Sine Wave")
plt.xlabel("x")
plt.ylabel("sin(x)")
plt.grid()
plt.show()
````
