```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
data = np.loadtxt("luftsensor_studie8.txt", skiprows=1, delimiter=',')
duration = data[:, 0]
sensor = data[:, 1]
referenz = data[:, 2]
```

# Teil 1

## Frage 1

```python
unique_durations = np.unique(duration)
y_vals = []

for d in unique_durations:
    mask = (duration == d)
    total = np.sum(mask)
    correct = np.sum(sensor[mask] == referenz[mask])
    y_vals.append(correct / total)
    # print(correct / total)

X = unique_durations
Y = np.log(1 - np.array(y_vals))
# print(Y)

b, ln_a = np.polyfit(X, Y, 1)
a = np.exp(ln_a)

print(f"a = {a:.5f}")
print(f"b = {b:.5f} 1/min")
```

## Frage 2

```python
x1 = 3.0
y1 = 1 - a * np.exp(b * x1)
y1_percent = y1 * 100
print(f"probability @ 3min = {y1_percent:.5f}%")

x_min = np.log((1 - 0.85) / a) / b
print(f"Minimum time needed for 85% accuracy = {x_min:.5f}min")
```

# Teil 2

## Frage 3

```python
p = y1
N = 2000

sigma_y = np.sqrt(p * (1 - p) / N)

untere_prozent = (p - 2*sigma_y) * 100
obere_prozent = (p + 2*sigma_y) * 100

print(f"sigma = {sigma_y:.5f}")
```

## Frage 4

```python
print(f"Untere Grenze = {untere_prozent:.5f}%")
print(f"Obere Grenze = {obere_prozent:.5f}%")
```

# Teil 3

## Frage 5

```python
x_val = 3.0
sigma_x = 0.2

dy_dx = -a * b * np.exp(b * x_val)   # wird positiv, weil b negativ
sigma_y_extra = abs(dy_dx) * sigma_x
fehler_prozentpunkte = sigma_y_extra * 100

print(f"{fehler_prozentpunkte:.5f}%")
```

# Teil 4

## Frage 6

```python
p = y1
P_K = 0.25
P_nK = 1 - 0.25

P_K_given_M = (p * P_K) / (p * P_K + (1 - p) * P_nK)
print(f"{(P_K_given_M * 100):.5f}%")
```

## Frage 7

```python
p2 = p**2
p_falsch2 = (1 - p)**2

P_K_given_M2 = (p2 * P_K) / (p2 * P_K + p_falsch2 * P_nK)
print(f"{(P_K_given_M2 * 100):.5f}%")
```
