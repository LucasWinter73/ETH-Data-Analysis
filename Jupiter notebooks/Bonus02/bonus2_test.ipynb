```python
import numpy as np
```

```python
data = np.loadtxt('luftsensor_studie8.csv', skiprows=1, delimiter=',')
x_min = data[:, 0]
sensor = data[:, 1]
referenz = data[:, 2]

unique_durations = np.unique(x_min)
y = []

for i in unique_durations:
    mask = (x_min == i)
    total = np.sum(mask)
    correct = np.sum(sensor[mask] == referenz[mask])
    y.append(correct / total)

b_fit, ln_a = np.polyfit(unique_durations, np.log(1 - np.array(y)), 1)
a_fit = np.exp(ln_a)

print(f"a = {a_fit:.5f}")
print(f"b = {b_fit:.5f} 1/min")
```

y = 1 - a*e^(bx)
(1 - y) / a = e^(bx)
ln((1 - y) / a) = bx
ln((1 - y) / a) / b = x

```python
def y_at(x):
    return 1 - a_fit * np.exp(b_fit * x)

print(f"y(x = 3min) = {y_at(3) * 100}%")

def x_at(y):
    return np.log((1 - y) / a_fit) / b_fit

print(f"x(y = 85%) = {x_at(0.85)} min")
```

```python
sigma = np.sqrt(y_at(3) * (1 - y_at(3)) / 2000)

print(f"sigma = {sigma}")

untere_prozent = (y_at(3) - 2*sigma) * 100
obere_prozent = (y_at(3) + 2*sigma) * 100
print(f"Untere Grenze = {untere_prozent:.5f}%")
print(f"Obere Grenze = {obere_prozent:.5f}%")
```

```python
x_val = 3.0
sigma_x = 0.2

dy_dx = -a_fit * b_fit * np.exp(b_fit * x_val)
print(f"{dy_dx * sigma_x * 100:.5f}%")
```

P(k|s) = P(s|k) * P(k) / P(s)
P(k) = 0.25
P(s) = y_at(3)

```python
p_k = 0.25
p_s = y_at(3)

p_k_s = (p_s * p_k) / ((p_s * p_k) + ((1 - p_s) * (1 - p_k)))

print(f"{p_k_s * 100:.5f}%")
```

```python
p_k_s_2 = (p_s * p_k_s) / ((p_s * p_k_s) + ((1 - p_s) * (1 - p_k_s)))

print(f"{p_k_s_2 * 100:.5f}%")
```
